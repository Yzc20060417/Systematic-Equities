## 4.1.1 Helpers

```python
# Helpers
def _simplify_columns(df: pd.DataFrame) -> pd.DataFrame:
    """If columns are MultiIndex with constant metadata levels, drop those levels."""
    out = df.copy()
    if isinstance(out.columns, pd.MultiIndex):
        keep_levels = []
        for i in range(out.columns.nlevels):
            if out.columns.get_level_values(i).nunique() > 1:
                keep_levels.append(i)
        if len(keep_levels) == 1:
            out.columns = out.columns.get_level_values(keep_levels[0])
        else:
            out.columns = pd.Index(["|".join(map(str, tup)) for tup in out.columns.to_list()])
    out.columns = pd.Index([str(c).strip() for c in out.columns])
    return out

def _ensure_datetime_index(df: pd.DataFrame) -> pd.DataFrame:
    out = df.copy()
    if not isinstance(out.index, pd.DatetimeIndex):
        out.index = pd.to_datetime(out.index)
    out = out[~out.index.duplicated(keep="last")]
    out = out.sort_index()
    return out

def _to_numeric(df: pd.DataFrame) -> pd.DataFrame:
    out = df.copy()
    for c in out.columns:
        out[c] = pd.to_numeric(out[c], errors="coerce")
    out = out.replace([np.inf, -np.inf], np.nan)
    return out

def _cs_winsorize(df: pd.DataFrame, lo=0.01, hi=0.99, min_n=10) -> pd.DataFrame:
    """Winsorize cross-sectionally per date."""
    def _clip_row(x):
        x = x.astype(float)
        m = x.notna()
        if m.sum() < min_n:
            return x
        ql = x[m].quantile(lo)
        qh = x[m].quantile(hi)
        return x.clip(ql, qh)
    return df.apply(_clip_row, axis=1)

def _cs_zscore(df: pd.DataFrame, min_n=10) -> pd.DataFrame:
    """Z-score cross-sectionally per date (mean/std on available names)."""
    def _zrow(x):
        x = x.astype(float)
        m = x.notna()
        if m.sum() < min_n:
            return x * np.nan
        mu = x[m].mean()
        sd = x[m].std(ddof=0)
        if sd == 0 or np.isnan(sd):
            return x * np.nan
        out = (x - mu) / sd
        return out
    return df.apply(_zrow, axis=1)

def _cs_rank(df: pd.DataFrame, min_n=10, method="average") -> pd.DataFrame:
    """Cross-sectional ranks scaled to [-0.5, 0.5] (robust for small universes)."""
    def _rrow(x):
        x = x.astype(float)
        m = x.notna()
        if m.sum() < min_n:
            return x * np.nan
        r = x[m].rank(method=method, ascending=True)
        # map ranks to (0,1), then center to (-0.5,0.5)
        u = (r - 0.5) / m.sum()
        out = pd.Series(np.nan, index=x.index, dtype=float)
        out.loc[m] = u - 0.5
        return out
    return df.apply(_rrow, axis=1)
```

```python
# 1) Collect alpha pool
alpha_names = (
    ["alpha_nltsmom"]
    + [f"alpha{i}" for i in range(1, 25) if i not in (14, 20, 23)]
)

alpha_pool = {}
missing = []
wrong_type = []

for name in alpha_names:
    if name not in globals():
        missing.append(name)
        continue
    obj = globals()[name]
    if not isinstance(obj, pd.DataFrame):
        wrong_type.append((name, type(obj)))
        continue
    alpha_pool[name] = obj

print(f"Requested {len(alpha_names)} alphas.")
print(f"Collected  {len(alpha_pool)} DataFrames.")

if missing:
    print("Missing variables:", missing)
if wrong_type:
    print("Wrong type (expected pd.DataFrame):", wrong_type)
```

```python
# 3) Define master calendar + ticker set (using px_close)
# -----------------------------
# Set this once if your variable name differs:
# px_close = <YOUR CLOSE DF>
px_close = _simplify_columns(_ensure_datetime_index(_to_numeric(close_df)))

tickers_common = set(px_close.columns)
for k, df in alpha_pool.items():
    tickers_common &= set(df.columns)

tickers_common = sorted(tickers_common)
print(f"Common tickers across prices + all alphas: {len(tickers_common)}")

master_index = px_close.index  # use price calendar as source of truth

# align all alphas to master index + common tickers
alpha_aligned = {}
for k, df in alpha_pool.items():
    alpha_aligned[k] = df.reindex(index=master_index, columns=tickers_common)

px_close = px_close.reindex(index=master_index, columns=tickers_common)

# tradable universe mask (minimum): we can only trade names with a close price
univ = px_close.notna()
```

```python
# 4) Coverage report
# -----------------------------
def build_coverage_table(alpha_aligned: dict, univ: pd.DataFrame) -> pd.DataFrame:
    rows = []
    idx = univ.index
    for k, df in alpha_aligned.items():
        m = df.notna() & univ
        any_day = m.any(axis=1)
        start = any_day.idxmax() if any_day.any() else pd.NaT
        rows.append({
            "alpha": k,
            "start_date_nonnull": start,
            "end_date": idx.max(),
            "avg_names_per_day": float(m.sum(axis=1).mean()),
            "median_names_per_day": float(m.sum(axis=1).median()),
            "overall_coverage_frac": float(m.sum().sum() / univ.sum().sum()),
        })
    return pd.DataFrame(rows).sort_values("overall_coverage_frac", ascending=False).reset_index(drop=True)

coverage = build_coverage_table(alpha_aligned, univ)
display(coverage)

# Optional: quick sanity checks
assert px_close.index.is_monotonic_increasing
for k, df in alpha_aligned.items():
    assert df.index.equals(master_index), f"{k} index misaligned"
    assert list(df.columns) == tickers_common, f"{k} columns misaligned"
```

```python
# 5) Robust cross-sectional cleaning (analysis-ready exposures)
# -----------------------------
alpha_clip_z = {}
alpha_rank = {}

for k, df in alpha_aligned.items():
    # keep values only where tradable
    x = df.where(univ)

    # winsorize then z-score (classic)
    xw = _cs_winsorize(x, lo=0.01, hi=0.99, min_n=10)
    xz = _cs_zscore(xw, min_n=10)

    # rank version (often best for STOXX50)
    xr = _cs_rank(x, min_n=10)

    alpha_clip_z[k] = xz
    alpha_rank[k] = xr

print("Built cleaned alpha versions:")
print(" - alpha_aligned: raw aligned")
print(" - alpha_clip_z: winsorized + cross-sectional z-score")
print(" - alpha_rank: cross-sectional ranks in [-0.5, 0.5]")
```

```python
# 1) Choose factor subsets by coverage thresholds
# ---------------------------
def select_alphas_by_coverage(coverage: pd.DataFrame,
                             min_overall=0.60,
                             min_median_names=30) -> list[str]:
    m = (coverage["overall_coverage_frac"] >= min_overall) & (coverage["median_names_per_day"] >= min_median_names)
    return coverage.loc[m, "alpha"].tolist()

alphas_core = select_alphas_by_coverage(coverage, min_overall=0.80, min_median_names=45)   # high-quality set
alphas_pool = select_alphas_by_coverage(coverage, min_overall=0.55, min_median_names=25)   # broad pool (like your ~0.56 group)

print("Core set (high coverage):", alphas_core)
print("Broad pool (looser thresholds):", alphas_pool)

# ---------------------------
# 2) Build a "common-live" date mask
# ---------------------------
def common_date_mask(alpha_aligned: dict,
                     univ: pd.DataFrame,
                     alphas: list[str],
                     min_names_per_alpha_per_day=30,
                     require_frac_alphas=0.9) -> pd.Series:
    """
    For each day, check each alpha has >= min_names_per_alpha_per_day non-missing tradable names.
    Keep day if at least require_frac_alphas of alphas satisfy.
    """
    ok_matrix = []
    for k in alphas:
        m = alpha_aligned[k].notna() & univ
        ok = (m.sum(axis=1) >= min_names_per_alpha_per_day)
        ok_matrix.append(ok.rename(k))
    ok_df = pd.concat(ok_matrix, axis=1)
    frac_ok = ok_df.mean(axis=1)
    return frac_ok >= require_frac_alphas

# Example: common sample for the broad pool (tune thresholds)
mask_common = common_date_mask(alpha_aligned, univ, alphas=alphas_pool,
                               min_names_per_alpha_per_day=30,
                               require_frac_alphas=0.9)

print("Common-sample days:", int(mask_common.sum()), "out of", len(mask_common))
if mask_common.any():
    print("Common-sample start:", mask_common.idxmax())
    print("Common-sample end:", mask_common[mask_common].index.max())

# ---------------------------
# 3) Define explicit window starts based on latest factor "turn-on"
# ---------------------------
def window_from_latest_start(coverage: pd.DataFrame, alphas: list[str]) -> pd.Timestamp:
    s = coverage.set_index("alpha").loc[alphas, "start_date_nonnull"]
    return pd.to_datetime(s.max())

latest_start_pool = window_from_latest_start(coverage, alphas_pool)
latest_start_core = window_from_latest_start(coverage, alphas_core)

print("Latest start (broad pool):", latest_start_pool)
print("Latest start (core set):", latest_start_core)
```

```python
# - alpha_rank: dict[name -> df]  (cross-sectional ranks, aligned)
# - alpha_clip_z: dict[name -> df] (winsorized + cs zscore, aligned)
# - px_close: DataFrame of close prices aligned to same index/columns
# - univ: boolean tradable mask, typically px_close.notna()
# - mask_common: boolean Series on dates (your common-sample date mask)
# - alphas_pool (or alphas_core): list[str] of factor names to include

ALPHAS = alphas_pool          # <-- choose: alphas_core or alphas_pool
DATE_MASK = mask_common       # <-- your chosen date mask
H = 1                         # forward horizon in days (set 5, 20 later if needed)

# ---------------------------
# 1) Forward returns (simple close-to-close)
# ---------------------------
# R_fwd is indexed by "signal date" t, whose return is from t -> t+H
R_fwd = px_close.pct_change(H).shift(-H)
R_fwd = R_fwd.where(univ)

# Restrict to chosen dates
R_fwd = R_fwd.loc[DATE_MASK]

# ---------------------------
# 2) Stack factors into a single panel (MultiIndex columns)
# ---------------------------
def stack_alpha_dict(alpha_dict: dict, alphas: list[str], date_mask: pd.Series) -> pd.DataFrame:
    frames = []
    for a in alphas:
        df = alpha_dict[a].loc[date_mask]  # date-filtered
        # columns -> MultiIndex (alpha, ticker)
        df.columns = pd.MultiIndex.from_product([[a], df.columns], names=["alpha", "ticker"])
        frames.append(df)
    X = pd.concat(frames, axis=1).sort_index(axis=1)
    return X

X_rank = stack_alpha_dict(alpha_rank, ALPHAS, DATE_MASK)
X_z    = stack_alpha_dict(alpha_clip_z, ALPHAS, DATE_MASK)

print("X_rank shape:", X_rank.shape, "  (#dates x (#alphas * #tickers))")
print("X_z    shape:", X_z.shape)
print("R_fwd  shape:", R_fwd.shape)

# ---------------------------
# 3) Build a sample mask requiring sufficient data per day
# ---------------------------
# For each day and ticker, require:
# - tradable (univ True)
# - returns available
# - at least K factors non-missing (so regression/corr isn't junk)
K_MIN_FACTORS = max(5, int(0.7 * len(ALPHAS)))   # e.g. require 70% of factors available

# Count how many factor exposures are available per (date,ticker)
# reshape X_rank to (date, alpha, ticker) via stack/unstack
avail = X_rank.notna()
# convert MultiIndex columns -> count per ticker
avail_counts = avail.groupby(level="ticker", axis=1).sum()  # sum over alpha columns per ticker
# avail_counts: (date x ticker) = #non-missing alphas

sample_mask = (univ.loc[DATE_MASK] &
               R_fwd.notna() &
               (avail_counts >= K_MIN_FACTORS))

print("K_MIN_FACTORS =", K_MIN_FACTORS)
print("Avg names/day in sample:", float(sample_mask.sum(axis=1).mean()))
print("Median names/day in sample:", float(sample_mask.sum(axis=1).median()))

# Apply sample mask to returns
R_fwd_s = R_fwd.where(sample_mask)

# Quick diagnostics
bad_days = sample_mask.sum(axis=1).sort_values().head(10)
display(bad_days.rename("names_in_sample_bottom10"))
```

```python
ALPHAS = alphas_pool        # or alphas_core
DATE_MASK = mask_common
MIN_NAMES = 30              # same scale as your sample rules
MIN_CS_STD = 1e-6           # for rank panel; use larger (e.g. 1e-3) for z panel

def factor_viability_mask(alpha_dict, univ, alphas, date_mask, min_names=30, min_cs_std=1e-6):
    """
    Returns:
      viable[a]: Series[date] whether factor a is usable on that date
      stats: DataFrame with viability summary per factor
    """
    viable = {}
    rows = []

    for a in alphas:
        x = alpha_dict[a].loc[date_mask].where(univ.loc[date_mask])
        n = x.notna().sum(axis=1)
        cs_std = x.std(axis=1, skipna=True)

        v = (n >= min_names) & (cs_std > min_cs_std)
        viable[a] = v

        rows.append({
            "alpha": a,
            "viable_day_frac": float(v.mean()),
            "avg_names_when_viable": float(n[v].mean()) if v.any() else np.nan,
            "median_names_when_viable": float(n[v].median()) if v.any() else np.nan,
            "median_cs_std": float(cs_std.median()),
            "zero_std_day_frac": float((cs_std.fillna(0) == 0).mean()),
        })

    stats = pd.DataFrame(rows).sort_values("viable_day_frac", ascending=False).reset_index(drop=True)
    return viable, stats

viable_rank, viable_stats_rank = factor_viability_mask(
    alpha_rank, univ, ALPHAS, DATE_MASK, min_names=MIN_NAMES, min_cs_std=MIN_CS_STD
)

display(viable_stats_rank)
```

## 4.2.1 Spearman Correlation

```python
from itertools import combinations

def daily_cs_corr(xi: pd.DataFrame, xj: pd.DataFrame, min_names=30):
    """
    For each day: corr across tickers between xi[t,:] and xj[t,:]
    Returns: Series indexed by date
    """
    # align
    xi, xj = xi.align(xj, join="inner", axis=0)
    corrs = []
    dates = xi.index
    for t in dates:
        a = xi.loc[t]
        b = xj.loc[t]
        m = a.notna() & b.notna()
        if m.sum() < min_names:
            corrs.append(np.nan)
        else:
            corrs.append(a[m].corr(b[m], method="spearman"))
    return pd.Series(corrs, index=dates)

def factor_corr_matrix(alpha_rank, univ, alphas, date_mask, viable=None, min_names=30):
    alphas = list(alphas)
    C = pd.DataFrame(np.nan, index=alphas, columns=alphas)
    N = pd.DataFrame(0, index=alphas, columns=alphas, dtype=int)

    # pre-mask
    X = {a: alpha_rank[a].loc[date_mask].where(univ.loc[date_mask]) for a in alphas}

    for a in alphas:
        C.loc[a, a] = 1.0
        N.loc[a, a] = int(date_mask.sum())

    for a, b in combinations(alphas, 2):
        xi, xj = X[a], X[b]

        # apply viability intersection if provided
        if viable is not None:
            v = viable[a].loc[date_mask] & viable[b].loc[date_mask]
            xi = xi.loc[v]
            xj = xj.loc[v]

        s = daily_cs_corr(xi, xj, min_names=min_names).dropna()
        C.loc[a, b] = C.loc[b, a] = float(s.mean()) if len(s) else np.nan
        N.loc[a, b] = N.loc[b, a] = int(len(s))

    return C, N

C_spear, N_eff = factor_corr_matrix(
    alpha_rank, univ, ALPHAS2, DATE_MASK, viable=viable_rank, min_names=MIN_NAMES
)

display(C_spear)
display(N_eff)
```

```python
C = C_spear.copy()
N = N_eff.copy()

# --- Factor-level "history quality": how many days do I share with others?
factor_min_N = N.replace(0, np.nan).min(axis=1)
factor_med_N = N.replace(0, np.nan).median(axis=1)

summary_N = pd.DataFrame({
    "min_pairwise_days": factor_min_N,
    "median_pairwise_days": factor_med_N,
}).sort_values("min_pairwise_days")

display(summary_N)

# --- Extract top correlated pairs above a threshold
def top_pairs(C, N, min_abs_corr=0.3, min_days=800, topk=30):
    alphas = list(C.index)
    out = []
    for i in range(len(alphas)):
        for j in range(i+1, len(alphas)):
            a, b = alphas[i], alphas[j]
            cij = C.loc[a, b]
            nij = N.loc[a, b]
            if pd.isna(cij): 
                continue
            if abs(cij) >= min_abs_corr and nij >= min_days:
                out.append((a, b, float(cij), int(nij)))
    out.sort(key=lambda x: abs(x[2]), reverse=True)
    return pd.DataFrame(out, columns=["alpha_i","alpha_j","mean_spearman","N_days"]).head(topk)

display(top_pairs(C, N, min_abs_corr=0.3, min_days=800, topk=50))
```

## Eigenspectrum

```python
# Use only factors with a fully defined correlation row/col
good = C.notna().all(axis=1)
C0 = C.loc[good, good].copy()

# If any tiny NaNs remain (shouldn't), fill with 0 as a conservative default
C0 = C0.fillna(0.0)

# Ensure symmetry
C0 = (C0 + C0.T) / 2

# Eigenvalues
eigvals = np.linalg.eigvalsh(C0.values)
eigvals_sorted = np.sort(eigvals)[::-1]

# Condition number (avoid division by 0)
eps = 1e-12
cond = float(eigvals_sorted[0] / max(eigvals_sorted[-1], eps))

# Effective rank (entropy-based)
p = np.clip(eigvals_sorted, 0, None)
p = p / (p.sum() + eps)
eff_rank = float(np.exp(-(p * np.log(p + eps)).sum()))

print(f"#factors used: {C0.shape[0]}")
print(f"min eig: {eigvals_sorted[-1]:.6g}  max eig: {eigvals_sorted[0]:.6g}")
print(f"condition number: {cond:.2f}")
print(f"effective rank: {eff_rank:.2f}")

plt.figure()
plt.plot(eigvals_sorted, marker="o")
plt.title("Eigenvalues of factor correlation matrix")
plt.xlabel("Index (sorted desc)")
plt.ylabel("Eigenvalue")
plt.show()
```

## 4.2.2 Returns Regression: Fama-Macbeth

```python
# Use z-scored exposures for regression
# You already have X_z (date x (alpha,ticker)) and R_fwd_s (date x ticker) and sample_mask (date x ticker)
# Ensure these correspond to the same DATE_MASK window already applied.
X_use = X_z.copy()
R_use = R_fwd_s.copy()
M_use = sample_mask.copy()

# --- utilities
def _row_to_matrix(X_row: pd.Series) -> pd.DataFrame:
    """X_row is a Series indexed by MultiIndex (alpha, ticker). Return DataFrame index=ticker, cols=alpha."""
    # X columns are (alpha, ticker), so X_row index is (alpha,ticker)
    # unstack alpha -> index=ticker, cols=alpha
    return X_row.unstack("alpha").sort_index(axis=1)

def fama_macbeth_factor_returns(X_panel: pd.DataFrame,
                               R_fwd: pd.DataFrame,
                               sample_mask: pd.DataFrame,
                               alphas: list[str],
                               add_intercept: bool = True,
                               min_nobs: int = 30,
                               fill_missing_exposure_with_zero: bool = True):
    """
    Estimate daily factor returns via cross-sectional OLS per day.
    X_panel: DataFrame indexed by date, columns MultiIndex (alpha,ticker)
    R_fwd:   DataFrame indexed by date, columns ticker
    sample_mask: boolean DataFrame indexed by date, columns ticker
    """
    dates = X_panel.index.intersection(R_fwd.index).intersection(sample_mask.index)
    betas = []
    meta = []

    for t in dates:
        y = R_fwd.loc[t]
        m = sample_mask.loc[t] & y.notna()
        if m.sum() < min_nobs:
            continue

        X_mat = _row_to_matrix(X_panel.loc[t])  # ticker x alpha
        # restrict to our alpha list and tickers in mask
        X_mat = X_mat.reindex(index=y.index, columns=alphas)
        X_mat = X_mat.loc[m.index[m]]  # only masked tickers
        y_t = y.loc[m.index[m]].astype(float)

        # handle missing exposures
        if fill_missing_exposure_with_zero:
            X_mat = X_mat.fillna(0.0)
        else:
            good = X_mat.notna().all(axis=1)
            X_mat = X_mat.loc[good]
            y_t = y_t.loc[good]

        if len(y_t) < min_nobs:
            continue

        # design matrix
        X_np = X_mat.values
        if add_intercept:
            X_np = np.column_stack([np.ones(len(y_t)), X_np])

        # OLS
        beta, *_ = np.linalg.lstsq(X_np, y_t.values, rcond=None)

        if add_intercept:
            alpha0 = beta[0]
            b = beta[1:]
        else:
            alpha0 = np.nan
            b = beta

        # R^2
        y_hat = X_np @ beta
        ss_res = np.sum((y_t.values - y_hat) ** 2)
        ss_tot = np.sum((y_t.values - y_t.values.mean()) ** 2)
        r2 = 1.0 - ss_res / ss_tot if ss_tot > 0 else np.nan

        betas.append(pd.Series(b, index=alphas, name=t))
        meta.append({"date": t, "nobs": int(len(y_t)), "r2": float(r2), "intercept": float(alpha0)})

    B = pd.DataFrame(betas).sort_index()
    stats = pd.DataFrame(meta).set_index("date").sort_index()
    return B, stats

B, fm_stats = fama_macbeth_factor_returns(
    X_panel=X_use,
    R_fwd=R_use,
    sample_mask=M_use,
    alphas=ALPHAS2,
    add_intercept=True,
    min_nobs=30,
    fill_missing_exposure_with_zero=True,
)

print("Factor returns panel B shape:", B.shape)
display(fm_stats.describe())

# --- Factor-return correlation (time-series corr)
corr_beta = B.corr()
display(corr_beta)

# --- Simple significance summary (t-stats assuming iid; optional NW next)
mu = B.mean()
sd = B.std(ddof=1)
tstat = mu / (sd / np.sqrt(B.shape[0]))

summary = pd.DataFrame({"mean_beta": mu, "std_beta": sd, "tstat_iid": tstat}).sort_values("tstat_iid", ascending=False)
display(summary)
```

## 4.3.1 Combined

```python
ALPHAS2 = list(B.columns)

# ---------- 1) Newey–West t-stats (sanity)
def newey_west_tstat(series: pd.Series, L: int = 5):
    x = series.dropna().values
    T = len(x)
    if T < 10:
        return np.nan
    mu = x.mean()
    y = x - mu
    gamma0 = np.dot(y, y) / T
    var = gamma0
    for l in range(1, L + 1):
        w = 1 - l / (L + 1)
        gam = np.dot(y[l:], y[:-l]) / T
        var += 2 * w * gam
    se = np.sqrt(var / T) if var > 0 else np.nan
    return mu / se if (se is not None and se > 0) else np.nan

L_NW = 5  # H=1, so 5 is fine; if you later use H=5/20, set L_NW ~ H or 2H
nw_t = B.apply(lambda s: newey_west_tstat(s, L=L_NW), axis=0)
print("NW t-stats:")
display(nw_t.sort_values(ascending=False))

# ---------- 2) Optional sign alignment (makes interpretation consistent)
mu_full = B.mean()
sign = np.sign(mu_full).replace(0, 1)  # if exactly zero, leave sign = +1

B_sa = B.mul(sign, axis=1)  # factor returns sign-aligned so mean >= 0
# For exposures, apply same sign so "positive score" always means expected positive return
def apply_sign_to_X(X_panel: pd.DataFrame, sign: pd.Series) -> pd.DataFrame:
    X = X_panel.copy()
    # multiply each alpha block by sign[alpha]
    for a, s in sign.items():
        if a in X.columns.get_level_values("alpha"):
            X.loc[:, (a, slice(None))] = X.loc[:, (a, slice(None))] * float(s)
    return X

X_sa = apply_sign_to_X(X_z, sign.loc[ALPHAS2])

# ---------- 3) Rolling factor-space optimizer
def ridge_mv_weights(B_window: pd.DataFrame, ridge: float = 1e-4):
    """
    w ∝ (Sigma + ridge*I)^(-1) mu
    Returns normalized weights with sum(|w|)=1
    """
    mu = B_window.mean().values
    Sigma = np.cov(B_window.values, rowvar=False, ddof=1)
    K = Sigma.shape[0]
    Sigma_r = Sigma + ridge * np.eye(K)
    w = np.linalg.solve(Sigma_r, mu)
    # normalize for stability
    s = np.sum(np.abs(w))
    if s == 0 or np.isnan(s):
        return np.zeros_like(w)
    return w / s

LOOKBACK = 252     # 1y rolling estimation
RIDGE = 1e-3       # increase if weights look noisy; decrease if too flat

dates = B_sa.index
W = pd.DataFrame(index=dates, columns=ALPHAS2, dtype=float)

for t_idx in range(len(dates)):
    t = dates[t_idx]
    # use only past data
    start = max(0, t_idx - LOOKBACK)
    window = B_sa.iloc[start:t_idx]  # excludes current day
    if len(window) < 60:  # require some history
        continue
    w = ridge_mv_weights(window.dropna(), ridge=RIDGE)
    W.loc[t] = w

W = W.dropna(how="all")
print("Weights panel shape:", W.shape)

# ---------- 4) Build combined score S[t, ticker] from X_sa and W
def row_to_Xmat(X_row: pd.Series) -> pd.DataFrame:
    # X_row indexed by (alpha,ticker) -> ticker x alpha
    return X_row.unstack("alpha").reindex(columns=ALPHAS2)

scores = []
score_dates = W.index.intersection(X_sa.index).intersection(R_fwd_s.index)

for t in score_dates:
    w = W.loc[t].astype(float)
    if w.isna().any():
        continue
    Xmat = row_to_Xmat(X_sa.loc[t])
    # apply tradable mask and keep only tickers in sample
    m = sample_mask.loc[t] & R_fwd_s.loc[t].notna()
    Xmat = Xmat.loc[m.index[m]].fillna(0.0)
    s = Xmat.values @ w.values
    scores.append(pd.Series(s, index=Xmat.index, name=t))

S = pd.DataFrame(scores).sort_index()
print("Combined score panel S shape:", S.shape)

# ---------- 5) Evaluate: daily IC of combined score vs forward returns
def daily_ic(score_df: pd.DataFrame, ret_df: pd.DataFrame, method="spearman", min_names=30):
    ics = []
    for t in score_df.index.intersection(ret_df.index):
        s = score_df.loc[t]
        r = ret_df.loc[t]
        m = s.notna() & r.notna()
        if m.sum() >= min_names:
            ics.append((t, s[m].corr(r[m], method=method)))
    return pd.Series(dict(ics)).sort_index()

ic = daily_ic(S, R_fwd_s, method="spearman", min_names=30)
print("IC mean:", float(ic.mean()), "IC std:", float(ic.std()), "ICIR:", float(ic.mean()/ic.std()) if ic.std()>0 else np.nan)
display(ic.describe())
```
